```python
import math
class Vector:
    def __init__(self, components):
        self.components = list(components)
        self.dim = len(self.components)
        
    def __add__(self, vector):
        return Vector([a + b for a,b in zip(self.components,vector.components)])
    
    def __sub__(self, vector):
        return Vector([a - b for a,b in zip(self.components,vector.components)])
        
    def dot(self, vector):
        return sum(a*b for a,b in zip(self.components, vector.components))
    
    # L2 Norm
    # The distance from the corner O (0,0,...0)
    def magnitude(self):
        return sum(x**2 for x in self.components)**0.5
        
    def normalize(self):
        mag = self.magnitude()
        return Vector(a/mag for a in self.components)
            
    def cosine_similarity(self, vector):
        return (self.dot(vector)/(self.magnitude()*vector.magnitude()))        
    def __repr__(self):
        return f"{self.components}"
```

```python
class Matrix:
    def __init__(self, rows):
        self.rows = [list (row) for row in rows]
        self.shape = (len(self.rows), len(self.rows[0]))
        
    def __matmul__(self, other):
        if isinstance(other,Vector):
            if other.dim != self.shape[1]:
                print("Unmatched")
                return
            return Vector(sum(self.rows[i][j]*other.components[j] for j in range(self.shape[1])) for i in range(self.shape[0]))
        if isinstance(other, Matrix):
            if other.shape[0] != self.shape[1]:
                print("Unmatched")
                return
            rows =[]
            for i in range(self.shape[0]):
                row = []
                for k in range(self.shape[1]):
                    num = 0
                    for l in range (other.shape[0]):
                        num += self.rows[i][k]*other.rows[l][k]
                        row.append(num)
                rows.append(row)
            return Matrix(rows)
        
    def transpose(self):
        rows =[]
        for i in range(self.shape[0]):
            row =[]
            for j in range (self.shape[1]):
                row.append(self.rows[j][i])
            rows.append(row)
        return Matrix(rows)
    
    def __repr__(self):
        return f"{self.rows}"
                    
        
```
